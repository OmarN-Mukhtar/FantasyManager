```python
# --- Same feature transformation as predictor.py, inlined ---
import sys, pathlib
_root = pathlib.Path.cwd()
_root = _root.parent if _root.name == "notebooks" else _root
if str(_root) not in sys.path:
    sys.path.insert(0, str(_root))

# Mirrors PlayerPredictor._build_features + _stack_horizons. Team/opponent ids here are
# built purely from club names seen in the CSV (predictor.py seeds them from the live
# bootstrap first); everything else matches.
from prediction.predictor import (
    BASE_ROLLING_FEATURES, ROLLING_WINDOWS, POINTS_LAGS, CONTEXT_FEATURES,
    TRAIN_LAST_N_GWS, RECENCY_DECAY, TRAIN_HORIZONS,
)

fdf = df.sort_values(["name", "season", "GW"]).copy()
fdf = fdf.groupby("name", group_keys=False).tail(TRAIN_LAST_N_GWS).copy()

grouped = fdf.groupby("name", sort=False)
feature_cols = []

# rolling means
for col in BASE_ROLLING_FEATURES:
    for w in ROLLING_WINDOWS:
        nm = f"{col}_rolling_{w}"
        fdf[nm] = grouped[col].transform(lambda s: s.rolling(window=w, min_periods=1).mean())
        feature_cols.append(nm)

# per-GW points lags (strictly prior GWs: shift(lag), not shift(lag-1))
for lag in range(1, POINTS_LAGS + 1):
    nm = f"points_gw_minus_{lag}"
    fdf[nm] = grouped["total_points"].shift(lag)
    feature_cols.append(nm)

# zero out defensive rolling stats for forwards
is_fwd = fdf["position"].astype(str).str.upper().eq("FWD")
if "element_type" in fdf.columns:
    is_fwd = is_fwd | (pd.to_numeric(fdf["element_type"], errors="coerce") == 4)
for base_col in ["goals_conceded", "clean_sheets"]:
    for w in ROLLING_WINDOWS:
        fdf.loc[is_fwd, f"{base_col}_rolling_{w}"] = 0.0

# stable club ids from names (FPL renumbers team ids each season)
club_names = set(fdf["team"].dropna()) | set(fdf["opp_team_name"].dropna())
team_name_to_id = {c: i + 1 for i, c in enumerate(sorted(club_names))}
fdf["team_id_feature"] = fdf["team"].map(team_name_to_id).fillna(0)
fdf["opponent_team_feature"] = fdf["opp_team_name"].map(team_name_to_id).fillna(0)

fdf["is_home_feature"] = pd.to_numeric(fdf.get("was_home", 0), errors="coerce").fillna(0)
fdr_col = fdf["fdr"] if "fdr" in fdf.columns else pd.Series(np.nan, index=fdf.index)
fdf["fdr_feature"] = pd.to_numeric(fdr_col, errors="coerce").fillna(3)
fdf["now_cost_feature"] = pd.to_numeric(fdf.get("now_cost", 0), errors="coerce").fillna(0)
fdf["availability_feature"] = pd.to_numeric(
    fdf.get("chance_of_playing_next_round", 100), errors="coerce"
).fillna(100) / 100.0

# stable per-player id
all_names = sorted(set(fdf["name"].dropna()))
player_stable_id = {n: i + 1 for i, n in enumerate(all_names)}
fdf["player_id_feature"] = fdf["name"].map(player_stable_id).fillna(0)

fdf["gameweeks_ahead"] = 0.0  # placeholder; real value set per-horizon below
feature_cols.extend(CONTEXT_FEATURES)

# recency-weighting ordinal
season_start_year = pd.to_numeric(fdf["season"].astype(str).str.slice(0, 4), errors="coerce")
fdf["season_ordinal"] = season_start_year * 38 + pd.to_numeric(fdf["GW"], errors="coerce")

# --- _stack_horizons: one row per (row, horizon h); target + fixture ctx pulled h GWs ahead ---
g = fdf.groupby("name", sort=False)
base_cols = [c for c in feature_cols if c != "gameweeks_ahead"]
parts = []
for h in TRAIN_HORIZONS:
    part = fdf[base_cols + ["season_ordinal"]].copy()
    part["target"] = g["total_points"].shift(-h)
    part["opponent_team_feature"] = g["opponent_team_feature"].shift(-h)
    part["is_home_feature"] = g["is_home_feature"].shift(-h)
    part["fdr_feature"] = g["fdr_feature"].shift(-h)
    part["availability_feature"] = g["availability_feature"].shift(-h)
    part["gameweeks_ahead"] = float(h)
    parts.append(part)

stacked = pd.concat(parts, ignore_index=True)[feature_cols + ["target", "season_ordinal"]].dropna()

X = stacked[feature_cols].astype(np.float32)
y = stacked["target"].astype(np.float32)
sample_weight = RECENCY_DECAY ** (stacked["season_ordinal"].max() - stacked["season_ordinal"])

print("engineered:", fdf.shape, "| stacked training frame:", stacked.shape)
stacked.head()
```

```python
# Forward-sum targets: total_points over the next h gameweeks starting from the current row
# (rows t..t+h-1), computed per (player, season) so nothing leaks across a season boundary.
# Inclusive of the current GW so h=38 from a GW1 row = whole rest of that player's season.
# Rows without a full h-GW window ahead (season end) become NaN and drop out of corr.
HORIZONS = [5, 10, 15, 20, 25, 30, 35, 38]

hdf = fdf.sort_values(["name", "season", "GW"]).copy()
gh = hdf.groupby(["name", "season"], sort=False)["total_points"]

def fwd_sum(s, h):
    # reverse -> trailing rolling sum of h -> reverse back = sum of rows i..i+h-1
    return s[::-1].rolling(h, min_periods=h).sum()[::-1]

target_cols = []
for h in HORIZONS:
    col = f"next_{h}gw_points"
    hdf[col] = gh.transform(lambda s, h=h: fwd_sum(s, h))
    target_cols.append(col)

# correlate every numeric feature (raw + engineered) against each horizon
num = hdf.select_dtypes("number")
feat_cols_for_corr = [c for c in num.columns if c not in target_cols]
corr_h = num.corr().loc[feat_cols_for_corr, target_cols]
corr_h = corr_h.reindex(corr_h[f"next_{HORIZONS[0]}gw_points"].abs().sort_values(ascending=False).index)

print("rows with a full window per horizon:")
print(hdf[target_cols].notna().sum())

plt.figure(figsize=(8, len(corr_h) * 0.22))
plt.imshow(corr_h, cmap="coolwarm", vmin=-1, vmax=1, aspect="auto")
plt.colorbar(fraction=0.046)
plt.xticks(range(len(target_cols)), target_cols, rotation=45, ha="right", fontsize=7)
plt.yticks(range(len(corr_h)), corr_h.index, fontsize=6)
plt.title("Feature correlation vs each forward-sum horizon")
plt.tight_layout()
plt.show()

corr_h.round(3)
```

```python
# === Three modelling frames: short (next game), medium (19 GW), long (38 GW) ===
# Features are per-GW-history quantities only (no bootstrap-snapshot columns like form /
# now_cost / ep_next / chance_of_playing, which are frozen across the season in the CSV).
# Snapshot equivalents rebuilt from history: pts_roll_4 (form), ppg_todate (points_per_game),
# value (price), selected (ownership). Fixture context is the NEXT fixture's (shift(-1)).

id_cols = ["name", "season", "GW", "position"]

base = fdf.sort_values(["name", "season", "GW"]).copy()
gps = base.groupby(["name", "season"], sort=False)

base["pts_roll_4"] = gps["total_points"].transform(lambda s: s.rolling(4, min_periods=1).mean())
base["ppg_todate"] = gps["total_points"].transform(lambda s: s.expanding().mean())
base["own_m"]      = pd.to_numeric(base["selected"], errors="coerce") / 1e6
_val = pd.to_numeric(base["value"], errors="coerce")
base["price"]      = np.where(_val > 30, _val / 10, _val)   # old rows in tenths, new in £M
for src, dst in [("is_home_feature", "is_home_next"),
                 ("fdr_feature", "fdr_next"),
                 ("opponent_team_feature", "opp_next")]:
    base[dst] = gps[src].shift(-1)

rate_core = [
    "pts_roll_4", "ppg_todate",
    "minutes_rolling_3", "minutes_rolling_5", "minutes_rolling_7", "starts",
    "bps_rolling_5", "bps_rolling_7",
    "ict_index_rolling_5", "ict_index_rolling_7",
    "expected_goal_involvements_rolling_5", "expected_goal_involvements_rolling_7",
    "expected_goals_conceded_rolling_7",
    "selected_rolling_5", "own_m", "price", "element_type",
]

short_feats = rate_core + [
    "minutes_rolling_1", "bps_rolling_3", "ict_index_rolling_3",
    "points_gw_minus_1", "points_gw_minus_2", "points_gw_minus_3",
    "points_gw_minus_4", "points_gw_minus_5",
    "is_home_next", "fdr_next", "opp_next",
    "transfers_balance",
]
medium_feats = rate_core
long_feats = [
    "pts_roll_4", "ppg_todate",
    "minutes_rolling_7", "starts", "bps_rolling_7", "ict_index_rolling_7",
    "expected_goal_involvements_rolling_7", "expected_goals_conceded_rolling_7",
    "selected_rolling_5", "own_m", "price", "element_type",
]


def horizon_points(frame, h, include_current):
    """Sum of total_points over the next h GWs, per (player, season).
    include_current=False -> rows t+1..t+h ; True -> rows t..t+h-1."""
    g = frame.groupby(["name", "season"], sort=False)["total_points"]
    s = g if include_current else g.shift(-1).groupby([frame["name"], frame["season"]], sort=False)
    return s.transform(lambda x: x[::-1].rolling(h, min_periods=h).sum()[::-1])


specs = {
    "short":  (short_feats,  1,  False),
    "medium": (medium_feats, 19, False),
    "long":   (long_feats,   38, True),   # inclusive so GW1 rows = whole remaining season
}

frames = {}
for tag, (feats, h, inc) in specs.items():
    feats = [c for c in feats if c in base.columns]
    out = base[id_cols + feats].copy()
    out[f"target_{h}gw"] = horizon_points(base, h, inc).values
    frames[tag] = out.dropna(subset=[f"target_{h}gw"]).reset_index(drop=True)

short_df, medium_df, long_df = frames["short"], frames["medium"], frames["long"]
for tag, f in frames.items():
    print(f"{tag:7s} {f.shape}  target={f.columns[-1]}")
short_df.head()
```

```python
# === Short-horizon bake-off: plain sklearn regressors, MSE loss, temporal split ===
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, HistGradientBoostingRegressor
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

d = short_df.copy()
d["is_home_next"] = d["is_home_next"].astype(float)
d["sortkey"] = pd.to_numeric(d["season"].str[:4]) * 38 + d["GW"]
feat_cols = [c for c in d.columns
             if c not in ("name", "season", "GW", "position", "target_1gw", "sortkey")]

CUT = 2025 * 38 + 18                       # train up to 2025-26 GW18
train = d[d["sortkey"] <= CUT]
test  = d[d["sortkey"] >= CUT + 2]         # test 2025-26 GW20+ (GW19 buffer)
Xtr, ytr = train[feat_cols], train["target_1gw"]
Xte, yte = test[feat_cols],  test["target_1gw"].to_numpy()
print(f"train {len(train)}  test {len(test)}")

pipe = lambda est: make_pipeline(SimpleImputer(strategy="median"), StandardScaler(), est)

models = {
    "LinearRegression": pipe(LinearRegression()),
    "Ridge":            pipe(Ridge(alpha=10.0)),
    "Lasso":            pipe(Lasso(alpha=0.01)),
    "RandomForest":     pipe(RandomForestRegressor(n_estimators=300, n_jobs=-1, random_state=0)),
    "GradBoost":        pipe(GradientBoostingRegressor(random_state=0)),
    "HistGradBoost":    HistGradientBoostingRegressor(random_state=0),
}

preds, rows = {}, []
for name, est in models.items():
    est.fit(Xtr, ytr)
    p = est.predict(Xte)
    preds[name] = p
    rows.append({
        "model": name,
        "RMSE":     round(mean_squared_error(yte, p) ** 0.5, 3),
        "MAE":      round(mean_absolute_error(yte, p), 3),
        "R2":       round(r2_score(yte, p), 3),
        "mean_pred": round(float(np.mean(p)), 2),
    })
display(pd.DataFrame(rows).sort_values("RMSE").reset_index(drop=True))

# ---- bias: does it over- or under-predict?  mean(pred - actual), overall and by actual bucket ----
buckets = pd.cut(yte, [-99, 0, 2, 5, 9, 99], labels=["0", "1-2", "3-5", "6-9", "10+"])
bias = {"overall": {}, "%over(pred>actual)": {}}
for b in buckets.categories:
    bias[f"y={b}"] = {}
for name, p in preds.items():
    resid = p - yte                                   # + = over-predict, - = under-predict
    bias["overall"][name] = round(resid.mean(), 3)
    bias["%over(pred>actual)"][name] = round(100 * (resid > 0).mean(), 1)
    for b in buckets.categories:
        m = buckets == b
        bias[f"y={b}"][name] = round(resid[m].mean(), 2)

print("\nmean(prediction - actual)   (+ over-predicts, - under-predicts)")
pd.DataFrame(bias)
```

## Component-simulation model for the upper tail

Don't regress `total_points`. Model the parts, each a clean sub-problem:

1. **minutes bucket** — P(0 / 1–59 / 60+), 3-class classifier
2. **goal rate** λ_g, **assist rate** λ_a — Poisson regressors (per appearance)
3. **clean sheet** P(CS | 60+) — classifier
4. **goals conceded** λ_gc — Poisson (for the −1 per 2, DEF/GK)
5. **saves** λ_sv — Poisson (GK only)
6. **bonus** — empirical lookup on (returns, CS)

Then Monte-Carlo each player-GW: draw components, apply the exact FPL scoring function,
get a full **points distribution**. From it: `E[pts]`, `P90` (ceiling), `P(pts ≥ 6)` (haul
probability). The tail is now a well-posed quantity instead of a regression target the mean
can never reach.


```python
# --- component training frame (built from `base`, which has raw per-GW cols + leak-free feats) ---
from sklearn.ensemble import HistGradientBoostingClassifier, HistGradientBoostingRegressor

cb = base.sort_values(["name", "season", "GW"]).copy()
gc = cb.groupby(["name", "season"], sort=False)
for col in ["minutes", "goals_scored", "assists", "clean_sheets",
            "goals_conceded", "saves", "bonus", "total_points"]:
    cb[f"nxt_{col}"] = gc[col].shift(-1)

# position (fall back to element_type for old rows with no position string)
et_map = {1: "GK", 2: "DEF", 3: "MID", 4: "FWD"}
cb["pos"] = cb["position"].where(cb["position"].isin(["GK", "DEF", "MID", "FWD"]),
                                 cb["element_type"].map(et_map))

feat_cols = [c for c in short_df.columns
             if c not in ("name", "season", "GW", "position", "target_1gw")]
cb["sortkey"] = pd.to_numeric(cb["season"].str[:4]) * 38 + cb["GW"]
cb["is_home_next"] = gc["is_home_feature"].shift(-1)          # (base only had own-row version here)
for src, dst in [("fdr_feature", "fdr_next"), ("opponent_team_feature", "opp_next")]:
    cb[dst] = gc[src].shift(-1)
cb["pts_roll_4"] = gc["total_points"].transform(lambda s: s.rolling(4, min_periods=1).mean())
cb["ppg_todate"] = gc["total_points"].transform(lambda s: s.expanding().mean())
cb["own_m"]      = pd.to_numeric(cb["selected"], errors="coerce") / 1e6
_v = pd.to_numeric(cb["value"], errors="coerce"); cb["price"] = np.where(_v > 30, _v / 10, _v)
cb["is_home_next"] = cb["is_home_next"].astype(float)
feat_cols = [c for c in feat_cols if c in cb.columns]

CUT = 2025 * 38 + 18
tr = cb[(cb["sortkey"] <= CUT) & cb["nxt_minutes"].notna()]
te = cb[cb["sortkey"] >= CUT + 2]
Xtr, Xte = tr[feat_cols], te[feat_cols]
print(f"train {len(tr)}  test {len(te)}")

def hgb_reg(**k):  return HistGradientBoostingRegressor(random_state=0, **k)

# 1. minutes bucket: 0 / 1-59 / 60+
mb_tr = np.select([tr.nxt_minutes.eq(0), tr.nxt_minutes.lt(60)], [0, 1], default=2)
mins_clf = HistGradientBoostingClassifier(random_state=0).fit(Xtr, mb_tr)

# 2-5. rate models, fit on appearances only (per-appearance rates)
app = tr[tr.nxt_minutes > 0]
Xapp = app[feat_cols]
lam_g  = hgb_reg(loss="poisson").fit(Xapp, app.nxt_goals_scored)
lam_a  = hgb_reg(loss="poisson").fit(Xapp, app.nxt_assists)
lam_gc = hgb_reg(loss="poisson").fit(Xapp, app.nxt_goals_conceded.clip(lower=0))
gk = app[app.pos == "GK"]
lam_sv = hgb_reg(loss="poisson").fit(gk[feat_cols], gk.nxt_saves) if len(gk) > 50 else None

# 3. clean sheet | played 60+
cs60 = tr[tr.nxt_minutes >= 60]
cs_clf = HistGradientBoostingClassifier(random_state=0).fit(
    cs60[feat_cols], (cs60.nxt_clean_sheets > 0).astype(int))

# 6. bonus: empirical mean by (returns bucket, cs) among appearances
app2 = tr[tr.nxt_minutes > 0].copy()
app2["ret"] = (app2.nxt_goals_scored + app2.nxt_assists).clip(0, 3).astype(int)
app2["csf"] = (app2.nxt_clean_sheets > 0).astype(int)
bonus_mean = app2.groupby(["ret", "csf"]).nxt_bonus.mean()
bonus_lut = np.array([[bonus_mean.get((r, c), 0.0) for c in (0, 1)] for r in range(4)])
print("bonus E by (returns,cs):\n", np.round(bonus_lut, 2))

# quick sanity on the rate heads
for nm, mdl, t in [("goals", lam_g, app.nxt_goals_scored),
                   ("assists", lam_a, app.nxt_assists),
                   ("gc", lam_gc, app.nxt_goals_conceded)]:
    print(f"{nm:8s} pred mean {mdl.predict(Xapp).mean():.3f}  actual mean {t.mean():.3f}")
```

```python
# --- Monte-Carlo the FPL scoring function over the drawn components ---
N = 1500
rng = np.random.default_rng(0)
n = len(te)

GOAL = te["pos"].map({"GK": 6, "DEF": 6, "MID": 5, "FWD": 4}).fillna(5).to_numpy()
CS   = te["pos"].map({"GK": 4, "DEF": 4, "MID": 1, "FWD": 0}).fillna(0).to_numpy()
is_dgk = te["pos"].isin(["GK", "DEF"]).to_numpy()
is_gk  = te["pos"].eq("GK").to_numpy()

Pm  = mins_clf.predict_proba(Xte)                       # (n,3) -> classes 0,1,2
cumP = np.cumsum(Pm, axis=1)
u = rng.random((n, N))
mbucket = (u[:, :, None] > cumP[:, None, :]).sum(axis=2)   # 0 / 1 / 2
played = mbucket > 0
p60    = mbucket == 2

lg  = np.clip(lam_g.predict(Xte),  0, None)[:, None]
la  = np.clip(lam_a.predict(Xte),  0, None)[:, None]
lgc = np.clip(lam_gc.predict(Xte), 0, None)[:, None]
lsv = np.clip(lam_sv.predict(Xte), 0, None)[:, None] if lam_sv is not None else np.zeros((n, 1))
pcs = cs_clf.predict_proba(Xte)[:, 1][:, None]

g  = rng.poisson(lg,  (n, N)) * played
a  = rng.poisson(la,  (n, N)) * played
gc_ = rng.poisson(lgc, (n, N)) * played
sv = rng.poisson(lsv, (n, N)) * played * is_gk[:, None]
cs = (rng.random((n, N)) < pcs) * p60

ret = np.clip(g + a, 0, 3)
csf = cs.astype(int)
bonus = bonus_lut[ret, csf] * played                      # expected-bonus proxy

pts = (np.where(mbucket == 2, 2, np.where(mbucket == 1, 1, 0))
       + g * GOAL[:, None] + a * 3 + cs * CS[:, None]
       + (sv // 3)
       - (gc_ // 2) * is_dgk[:, None]
       + bonus)
pts = np.where(played, pts, 0.0)

y   = te["nxt_total_points"].to_numpy()
gwx = te["sortkey"].to_numpy()
keep = ~np.isnan(y)
pts, y, gwx = pts[keep], y[keep], gwx[keep]
E   = pts.mean(1)
P90 = np.percentile(pts, 90, axis=1)
Phaul = (pts >= 6).mean(1)

# ---- eval ----
def bias_by_bucket(pred, tag):
    bk = pd.cut(y, [-99, 0, 2, 5, 9, 99], labels=["0", "1-2", "3-5", "6-9", "10+"])
    return pd.Series({b: round((pred - y)[bk == b].mean(), 2) for b in bk.categories}, name=tag)

def topk(pred, k=20):
    t = pd.DataFrame({"y": y, "p": pred, "g": gwx}); out = []
    for _, gg in t.groupby("g"):
        if len(gg) < k: continue
        best = gg.nlargest(k, "y").y.sum()
        if best > 0: out.append(gg.nlargest(k, "p").y.sum() / best)
    return round(float(np.mean(out)), 3)

from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
lin = make_pipeline(SimpleImputer(strategy="median"), StandardScaler(), LinearRegression()).fit(Xtr, tr.nxt_total_points)
lp = lin.predict(Xte)[keep]

print("mean(pred - actual) by actual bucket   (+ over, - under):")
display(pd.concat([bias_by_bucket(lp, "LinReg (MSE)"),
                   bias_by_bucket(E, "Sim E[pts]"),
                   bias_by_bucket(P90, "Sim P90")], axis=1))

print("\ntop-20 capture:")
print(f"  LinReg (MSE)     {topk(lp)}")
print(f"  Sim E[pts]       {topk(E)}")
print(f"  Sim P(pts>=6)    {topk(Phaul)}")
print(f"  Sim P90 ceiling  {topk(P90)}")

# calibration of the haul probability
cal = pd.DataFrame({"Phaul": Phaul, "hauled": (y >= 6).astype(int)})
cal["bin"] = pd.qcut(cal.Phaul, 10, duplicates="drop")
print("\nP(pts>=6) calibration:")
print(cal.groupby("bin", observed=True).agg(pred=("Phaul", "mean"),
                                            actual=("hauled", "mean"),
                                            n=("hauled", "size")).round(3).to_string())
```
